***
## DATA PREPROCESSING
***

```python
# Handling missing values

df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
# Feature Engineering

# Convert DATE features into Insights

# 1. age_of_customer
df["Age"] = 2026 - df["Year_Birth"]

# 2. customer_joining_date converted to no. of days from joining

df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst = True)

reference_date = df["Dt_Customer"].max()

df["Customer_Tenure_Days"] = (reference_date - df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
# Spending

df["Total_Spending"] = df["MntWines"] + df["MntFruits"] + df["MntFishProducts"] + df["MntMeatProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]

# Total Children
df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
df.head()
```

```python
df["Education"].value_counts()
```

```python
# convert to graduate, undergraduate, postgraduate

df["Education"] = df["Education"].replace({
    "Basic" : "Undergraduate",
    "2n Cycle": "Undergraduate",
    "Graduation" : "Graduate",
    "Master":"Postgraduate",
    "PhD":"Postgraduate"
})

df["Education"].value_counts()
```

```python
# Marital Status

df["Marital_Status"].value_counts()
```

```python
df["Living_With"] = df["Marital_Status"].replace({
    "Married" : "Partner",
    "Together": "Partner",
    "Single" : "Alone",
    "Widow":"Alone",
    "Divorced":"Alone",
    "Absurd" : "Alone",
    "YOLO" : "Alone"
})
```

```python
df["Living_With"].value_counts()
```

```python
# DROP UNNECESSARY COLUMNS

cols = ["ID","Marital_Status","Kidhome","Teenhome","Dt_Customer","Year_Birth"]
spending_cols = ["MntWines","MntFruits","MntFishProducts","MntMeatProducts","MntGoldProds","MntSweetProducts"]

cols_to_drop = cols + spending_cols

df_cleaned = df.drop(columns=cols_to_drop)
```

```python
df_cleaned.shape
```

```python
df_cleaned.head()
```

```python
# OUTLIER DETECTION

cols = ["Income", "Recency", "Response", "Age", "Total_Spending", "Total_Children"]

# relative plots of features - Pair Plot

sns.pairplot(df_cleaned[cols])
```

```python
# Remove Outliers

print("data size with outliers: ", len(df_cleaned))

df_cleaned = df_cleaned[(df_cleaned["Age"] < 90)]
df_cleaned = df_cleaned[(df_cleaned["Income"] < 600_000)]

print("data size without outliers: ", len(df_cleaned))
```

```python
# CORRELATION : HEATMAP

corr = df_cleaned.corr(numeric_only = True)

plt.figure(figsize = (8,6))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    annot_kws ={"size":6}
)
```

```python
# RED or Close - positive correlation
# BLUE or Close - negative correlation

# WHITE or close - no correlation
```

```python
# FEATURE ENCODING

from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder()

cat_cols = ["Education","Living_With"]

enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df = pd.DataFrame(enc_cols.toarray(), columns=ohe.get_feature_names_out(cat_cols), index = df_cleaned.index)
```

```python
enc_df.head()
```

```python
df_encoded = pd.concat([df_cleaned.drop(columns=cat_cols) , enc_df], axis=1)
```

```python
df_encoded.shape
```

```python
df_encoded.head()
```

```python
# FEATURE SCALING

from sklearn.preprocessing import StandardScaler

X = df_encoded
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
# DATA VIZUALIZATION

# 18 features so use PCA to reduce hence easy to vizualize

from sklearn.decomposition import PCA

pca = PCA(n_components = 2)
X_pca = pca.fit_transform(X_scaled)
```

```python
# plot

plt.scatter(X_pca[:,0], X_pca[:,1])
```

```python
pca.explained_variance_ratio_
# 23+11 = 34% captured
```

```python
pca = PCA(n_components = 3)
X_pca = pca.fit_transform(X_scaled)
```

```python
pca.explained_variance_ratio_
# 23+11+10 = 44% captured
```

```python
#plot
fig = plt.figure(figsize=(8,8))

ax= fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0], X_pca[:,1],X_pca[:,2])

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
```

```python
# IDENTIFY K VALUE
```

### ELBOW METHOD

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss=[]
for k in range(1,11):
    kmeans= KMeans(n_clusters=k, random_state = 42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
knee = KneeLocator(range(1,11),wcss,curve="convex",direction="decreasing")
optimal_k = knee.elbow
```

```python
print("best k =", optimal_k)
```

```python
# plot

plt.plot(range(1,11), wcss , marker='o')
```

### Silhouette Score

```python
from sklearn.metrics import silhouette_score

scores = []

for k in range(2,11):
    kmeans= KMeans(n_clusters=k, random_state = 42)
    labels = kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca, labels)
    scores.append(score)

#plot
plt.plot(range(2,11), scores , marker='o')
```

```python
# combined  plot

k_range = range(2,11)

fig , ax1 = plt.subplots(figsize=(8,6))

ax1.plot(k_range, wcss[:len(k_range)], marker="o", color="blue")

ax2 = ax1.twinx()
ax2.plot(k_range, scores[:len(k_range)], marker="x", color="red", linestyle="--")
```

```python
# with increase in silhouette WCSS decreases
#choose intersection k
```
